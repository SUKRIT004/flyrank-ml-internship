## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as “review every page with CTR below X” ignores search position and other signals and may produce many low-value recommendations. ML can combine multiple signals such as impressions, position, CTR, content type, age, trend, sessions, and engagement to rank opportunities more flexibly. The ML approach still needs to beat a transparent baseline on the chosen ranking metric; if it does not, the simpler rule should be preferred.
