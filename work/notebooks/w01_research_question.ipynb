## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 rows and 44 columns. After restricting the analysis to pages with impressions and a valid average position, 28,795 pages remained. The observed correlation between avg_position and CTR was -0.080, indicating a weak negative relationship. Mean CTR also varied across position tiers, from 2.764 in top_3 to 0.150 in deep. These descriptive differences make it worth investigating CTR relative to comparable search positions rather than using one global CTR threshold.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/SUKRIT004/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print(f"Dataset size: {len(df):,} rows × {len(df.columns)} columns")

visible = df[
    (df["impressions_90d"] > 0) &
    (df["avg_position"] > 0)
].copy()

correlation = visible["avg_position"].corr(visible["ctr"])

print(f"Visible pages with valid position: {len(visible):,}")
print(f"avg_position vs CTR correlation: {correlation:.3f}")

position_summary = (
    visible.groupby("position_tier")["ctr"]
    .mean()
    .sort_values(ascending=False)
)

print("\nMean CTR by position tier:")
print(position_summary)
```
